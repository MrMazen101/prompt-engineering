```python
import numpy as np
import matplotlib.pyplot as plt
import librosa
from scipy.signal import butter, filtfilt

# ---------------------------------------------------------
# 1) قراءة ملف الصوت باستخدام librosa (يدعم OGG)
# ---------------------------------------------------------
# غيّر الاسم هنا لو الملف مختلف
x, fs = librosa.load('freesound_community-water-dripping-ogg-format-70421.mp3', sr=None)

# تحويل الإشارة لـ float (librosa بالفعل بيرجع float)
x = x.astype(float)

# عدد العينات
N = len(x)

# محور الزمن
t = np.arange(N) / fs


# ---------------------------------------------------------
# 2) حساب Fourier Transform → X(f)
# ---------------------------------------------------------
X = np.fft.fft(x)                 # Fourier Transform
f = np.fft.fftfreq(N, 1/fs)       # محور الترددات


# ---------------------------------------------------------
# 3) حساب Amplitude Spectrum و Power Spectrum
# ---------------------------------------------------------
Amp = np.abs(X)                   # |X(f)|  ← Amplitude Spectrum
Power = Amp**2                    # |X(f)|^2 ← Power Spectrum


# ---------------------------------------------------------
# 4) إضافة نويز للإشارة
# ---------------------------------------------------------
noise = 0.2 * np.random.randn(N)  # White Gaussian Noise
x_noisy = x + noise               # الإشارة + النويز


# ---------------------------------------------------------
# 5) Spectrum للإشارة مع النويز
# ---------------------------------------------------------
Xn = np.fft.fft(x_noisy)
Amp_n = np.abs(Xn)
Power_n = Amp_n**2


# ---------------------------------------------------------
# 6) تصميم فلتر Low-pass لإزالة النويز
# ---------------------------------------------------------
cutoff = 4000                     # تردد القطع (Hz)
order = 4                         # درجة الفلتر

# تصميم فلتر Butterworth
b, a = butter(order, cutoff/(fs/2), btype='low')

# تطبيق الفلتر بدون Delay باستخدام filtfilt
x_filtered = filtfilt(b, a, x_noisy)


# ---------------------------------------------------------
# 7) Spectrum بعد الفلتر
# ---------------------------------------------------------
Xf = np.fft.fft(x_filtered)
Amp_f = np.abs(Xf)
Power_f = Amp_f**2


# ---------------------------------------------------------
# 8) رسم النتائج (Time Domain)
# ---------------------------------------------------------
plt.figure(figsize=(12,8))

plt.subplot(3,1,1)
plt.plot(t, x)
plt.title('Original Signal (Time Domain)')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')

plt.subplot(3,1,2)
plt.plot(t, x_noisy)
plt.title('Noisy Signal (Time Domain)')
plt.xlabel('Time (s)')

plt.subplot(3,1,3)
plt.plot(t, x_filtered)
plt.title('Filtered Signal (Time Domain)')
plt.xlabel('Time (s)')

plt.tight_layout()
plt.show()


# ---------------------------------------------------------
# 9) رسم النتائج (Frequency Domain)
# ---------------------------------------------------------
pos = f >= 0   # نعرض نصف الترددات الموجبة فقط

plt.figure(figsize=(12,8))

plt.subplot(3,1,1)
plt.plot(f[pos], Amp[pos])
plt.title('Original Amplitude Spectrum')
plt.xlabel('Frequency (Hz)')
plt.ylabel('|X(f)|')

plt.subplot(3,1,2)
plt.plot(f[pos], Amp_n[pos])
plt.title('Noisy Amplitude Spectrum')
plt.xlabel('Frequency (Hz)')

plt.subplot(3,1,3)
plt.plot(f[pos], Amp_f[pos])
plt.title('Filtered Amplitude Spectrum')
plt.xlabel('Frequency (Hz)')

plt.tight_layout()
plt.show()
```
